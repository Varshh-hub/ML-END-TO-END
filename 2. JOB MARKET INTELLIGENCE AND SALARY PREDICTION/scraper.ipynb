```python
import requests
import pandas as pd
import time
```

```python
APP_ID = "67334a16"
APP_KEY = "e78c57fd6ccc53aefea40098e9026670"
```

```python
def scrape_jobs(
    app_id,
    app_key,
    what,
    where="India",
    pages=10,
    results_per_page=50
):
    
    all_jobs = []

    for page in range(1, pages + 1):

        url = "https://api.adzuna.com/v1/api/jobs/in/search/{}".format(page)

        params = {
            "app_id": app_id,
            "app_key": app_key,
            "results_per_page": results_per_page,
            "what": what,
            "where": where,
            "content-type": "application/json"
        }

        response = requests.get(url, params=params, timeout=20)

        if response.status_code != 200:
            print(f"Page {page} failed: {response.status_code}")
            continue

        data = response.json()

        jobs = data.get("results", [])

        print(f"Page {page}: {len(jobs)} jobs collected")

        for job in jobs:

            all_jobs.append({
                "job_id": job.get("id"),
                "job_title": job.get("title"),
                "company": job.get("company", {}).get("display_name"),
                "location": job.get("location", {}).get("display_name"),
                "salary_min": job.get("salary_min"),
                "salary_max": job.get("salary_max"),
                "description": job.get("description"),
                "job_url": job.get("redirect_url"),
                "created": job.get("created")
            })

        time.sleep(1)

    return pd.DataFrame(all_jobs)
```

```python
roles = [
    # Data & AI
    "Data Scientist",
    "Data Analyst",
    "Machine Learning Engineer",
    "Data Engineer",
    "Business Analyst",
    "AI Engineer",
    "Business Intelligence Analyst",

    # Software Development
    "Software Developer",
    "Software Engineer",
    "Frontend Developer",
    "Backend Developer",
    "Full Stack Developer",
    "Web Developer",
    "Mobile App Developer",

    # Cloud & DevOps
    "Cloud Engineer",
    "DevOps Engineer",
    "Site Reliability Engineer",
    "Cloud Architect",
    "System Administrator",

    # Cybersecurity
    "Cybersecurity Engineer",
    "Cybersecurity Analyst",
    "Information Security Analyst",
    "Security Engineer",

    # Specialized AI / Data
    "AI/ML Researcher",
    "NLP Engineer",
    "Computer Vision Engineer",
    "MLOps Engineer",
    "Data Architect",
    "Solutions Architect"
]
```

```python
all_data = []
for role in roles:

    print(f"\nScraping: {role}")

    role_df = scrape_jobs(
        app_id=APP_ID,
        app_key=APP_KEY,
        what=role,
        where="India",
        pages=10,
        results_per_page=50
    )

    role_df["search_role"] = role
    all_data.append(role_df)

    time.sleep(2)
jobs_df = pd.concat(all_data, ignore_index=True)
```
